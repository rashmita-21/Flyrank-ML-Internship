### What I found in the starter dataset

The starter dataset contains **30,000 content items** across **32 clients**. Of these, **16,262 items have a downward trend**, which is approximately 54.2% of the dataset. The median number of impressions over 90 days is **731**.

However, a downward trend does not prove that a page needs a refresh or that editing it will improve performance.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

data_url = (
    "https://raw.githubusercontent.com/"
    "rashmita-21/Flyrank-ML-Internship/"
    "main/data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(data_url)

print("Dataset shape (rows, columns):", df.shape)
print("Unique content items:", df["content_id"].nunique())
print("Unique clients:", df["client_id"].nunique())

print(
    "Content items with a downward trend:",
    (df["trend_direction"] == "down").sum()
)

print(
    "Median impressions over 90 days:",
    df["impressions_90d"].median()
)

print("\nSample of performance data:")
display(
    df[
        [
            "impressions_90d",
            "sessions_90d",
            "content_age_days",
            "trend_direction",
        ]
    ].head()
)
```
